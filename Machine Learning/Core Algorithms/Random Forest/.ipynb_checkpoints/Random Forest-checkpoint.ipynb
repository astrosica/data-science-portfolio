# Predicting Loan Defaults with Random Forest

I predict loan default likelihood based on financial data using a random forest model.

## About the Data
I will use publicly available data from [LendingClub.com](www.lendingclub.com), a lending platform that connects borrowers with investors. 

The following summarizes the features in the data:
- **credit.policy**: whether the customer meets the credit criteria (1 if yes, 0 if no)
- **purpose**: the purpose of the loan (takes values "credit_card", "debt_consolidation", "educational", "major_purchase", "small_business", and "all_other")
- **int.rate**: the interest rate of the loan; borrowers deemed more risky are assigned higher interest rates
- **installment**: monthly installments owed by the borrower if the loan is funded
- **log.annual.inc**: natural log of the self-reported annual income of the borrower
- **dti**: debt-to-income ratio of the borrower
- **fico**: FICO credit score of the borrower
- **days.with.cr.line**: number of days the borrower has had a credit line
- **revol.bal**: borrower's revolving balance (amount unpaid at the end of the credit card billing cycle)
- **revol.util**: borrower's revolving line utilization rate (the amount of the credit line used relative to total credit available)
- **inq.last.6mths**: borrower's number of inquiries by creditors in the last 6 months
- **delinq.2yrs**: number of times the borrower has been 30+ days past due on a payment in the past 2 years
- **pub.rec**: borrower's number of derogatory public records (bankruptcy filings, tax liens, or judgments)
- **not.fully.paid**: whether the customer has failed to fully pay their loan (1 if yes, 0 if no)

## Step 5. Train Test Split
I'll follow the standard process of separating the input features (`X`) from the target variable (`y`) and splitting the data into training (`_train`) and testing (`_test`) sets using `scikit-learn`'s `train_test_split()` function. Splitting the data into training and testing sets allows us to evaluate the model performance using an independent dataset that the model hasn't seen during its training phase. This is an essential step for determining how well an ML model generalizes to unseen data.

```python
# define features
X = df.drop(columns=['not.fully.paid'])

# define target variable
y = df['not.fully.paid']
```

While creating the training and testing sets, I will use `train_test_split`'s `stratify` argument to ensure that the relative proportions of each class (negative and positive) are the same in all subsets as the original dataset. The `train_test_split()` function randomly splits the data into training and testing sets, so not using `stratify` could result in little to none of the minority class label in the training data, which would make it even more challenging for the model to learn from them.

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
```

## Step 6. Random Forest Modelling with Hyperparameter Tuning
Our data have been cleaned and pre-processed so I can start training a random forest model using `scikit-learn`'s `RandomForestClassifier()` function. I will start with hyperparameter tuning, after which I will incorporate SMOTE resampling in the next section to see if it improves recall on the minority class.

Since those who default on their loans only account for 16% of all data samples, I need to carefully consider what performance metrics we want to optimize. A model that is optimized for accuracy, for example, could simply predict the majority class label most of the time and still achieve a high accuracy. Instead, we want to prioritize **recall on the minority class label** since it is important to predict those who are likely to default on their loan. However, if I tune the hyperparameters by simply optimizing recall, we could be left with a model that similarly predicts all individuals to default on their loan, which would be misleading and unhelpful.

Instead, I will tune the hyperparameters by optimizing on average precision (AP), which is a common proxy for the area under the precision-recall curve (PR-AUC). The AP quantifies how well a model can discriminate between the positive and negative class across all thresholds. In this way, hyperparameter tuning allows us to find the general class of models that best differentiates between those who are likely to default and those who are not.

From here, I turn to threshold optimization to optimize the model for recall of the minority class. The threshold is the cutoff probability used by the model to differentiate between positive and negative class labels. By default, `RandomForestClassifier()` uses a threshold of 0.5, which is often not an ideal cutoff for imbalanced datasets. However, finding the threshold that maximizes recall would again result in a model that incorrectly predicts more people to default on their loan. Instead, I will optimize the F2 score, which is a variant of the F1 score that weighs recall more than precision. This combination of hyperparameter tuning and threshold optimization can therefore be used to tune a model that is optimized most heavily for recall while still considering the average precision.

### Step 6.1. Hyperparameter Tuning

Since we know that there is a signifncant class imbalance, I will use the `class_weight='balanced'` argument in `RandomForestClassifier()` to adjust the weights of each class to ensure that they contribute equally to the training phase. Since I will be fine-tuning a large number of parameters, I will set set `n_jobs=-1` to use all CPU cores and use `RandomizedSearchCV()` instead of `GridSearchCV()` for faster hyperparameter tuning over a larger parameter space. I will randomly search 100 samples of the parameter search grid using `n_iter=100`, use 3 equal folds in our cross-validation using `cv=3`, and optimize average precision (AP) using `scoring='average_precision'`. By default, this will assume that AP will be optimized for the `label=1` positive class label, which applies in our case.

The `RandomizedSearchCV()` function implicitly assumes k-fold cross validation, which randomly splits the data into k folds and could result in some folds having little to no samples from the minority class label. To ensure that all k folds have a representative proportion of class labels, I will use the `StratifiedKFold()` function for cross-validation.

```python
# define stratified k-fold cross validator
cv = StratifiedKFold(n_splits=3,shuffle=True,random_state=42)

# define parameter grid
param_grid = {
    'n_estimators': [50, 100, 300, 500],
    'max_depth': [10, 20, 50, None],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [10, 20, 50, 100, 200],
    'class_weight': ['balanced']
}

# build the random forest model
rf = RandomForestClassifier(random_state=42)

# randomized search optimizing recall
search_rf = RandomizedSearchCV(estimator=rf,param_distributions=param_grid,n_iter=100,cv=cv,scoring='average_precision',random_state=42,n_jobs=-1,verbose=1)

# fit the model
search_rf.fit(X_train,y_train)

# extract the model with the best recall
rf_tuned = search_rf.best_estimator_

# make predictions on the test set
y_pred_rf_tuned = rf_tuned.predict(X_test)

# print best-fit model parameters
print(f"best-fit model parameters (AP={search_rf.best_score_:.2f}): {search_rf.best_params_}")
```

Hyperparameter tuning yields a random forest model with an optimized AP of 27%. I will now turn to threshold optimization to improve recall.

### Step 6.2. Threshold Optimization
Now that the random forest model has been optimized for AP, I will find the threshold that optimizes recall. To do this, I will use the training set to create another train-test split (using the suffix `_tr`) so there is no data leakage into the testing set.

During threshold optimization I will optimize for the F2 score, which is a variation of the F1 score that weights recall more heavily than precision:

$$
\begin{align}
\text{F2 Score} = 5 \cdot \frac{\text{Precision} \cdot \text{Recall}}{4 \cdot \text{Precision} + \text{Recall}}.
\end{align}
$$

This allows us to primarily optimize recall while still considering the overall precision.

```python
# create another train-test split using the training data only
X_train_tr,X_test_tr,y_train_tr,y_test_tr = train_test_split(X_train, y_train, test_size=0.25, stratify=y_train, random_state=42)

# re-fit the model with the best-params model using the new training subset
rf_best = RandomForestClassifier(random_state=42,**search_rf.best_params_)
rf_best.fit(X_train_tr,y_train_tr)

# get predicted probabilities
y_proba = rf_best.predict_proba(X_test_tr)[:, 1]

# iterate through thresholds
thresholds = np.linspace(0,1,100)

# collect F2 scores
F2_scores = []
for threshold in thresholds:
    y_pred = (y_proba >= threshold).astype(int)
    F2_score = fbeta_score(y_test_tr, y_pred, beta=2)
    F2_scores.append(F2_score)

# find threshold that maximizes F2 score
idx = np.argmax(F2_scores)
best_threshold_rf_tuned = thresholds[idx]
best_F2_score = F2_scores[idx]

print(f"Best threshold: {best_threshold_rf_tuned:.4f}, Optimal F2 score: {best_F2_score:.3f}")
```

This yields an optimal threshold of 0.3636, which has a corresponding F2 score of 0.523. Next, I will train our final model on the entire training set and then tune the threshold to this optimal value. 

```python
# fit the model on the entire training set
rf_tuned = RandomForestClassifier(random_state=42,**search_rf.best_params_)
rf_tuned.fit(X_train,y_train)

# make predictions on entire testing set
y_proba_rf_tuned = rf_tuned.predict_proba(X_test)[:,1]

# make final predictions using optimized threshold
y_predict_rf_tuned = (y_proba_rf_tuned>best_threshold_rf_tuned).astype(int)
```

### Step 6.3. Evaluation Metrics
With our final random forest model trained, I will now use evaluation metrics to quantify its performance.

#### Step 6.3.1. Confusion Matrix
The confusion matrix is a diagnostic tool used to evaluate the performance of a classification model by summarizing how the predicted class labels compare to the actual class labels. For binary classification, the confusion matrix is a 2x2 matrix where the elements along the diagonal represent the true negative (TN) and true positive (TP) class labels (i.e., correct labels), while the off-diagonal elements represent the false negative (FN) and false positive (FP) class labels (i.e., incorrect labels). This simple structure demonstrates whether the model is able to properly separate positive and negative class labels, and whether there is any confusion between them.

It is worth noting that different sources use different conventions for the confusion matrix; `scikit-learn` uses the following convention: "...the count of true negatives is [0,0], false negatives is [1,0], true positives is [1,1] and false positives is [0,1]." Knowing this allows us to calculate two additional metrics that aren't included in the classification report (next step): the false positive rate (FPR) and false negative rate (FNR).

```python
# compute confusion matrix
cm_rf_tuned = confusion_matrix(y_test,y_predict_rf_tuned)
print(cm_rf_tuned)

# plot confusion matrix display
disp = ConfusionMatrixDisplay(confusion_matrix=cm_rf_tuned, display_labels=['Negative Class', 'Positive Class'])
disp.plot(cmap=plt.cm.Blues)
plt.title('Confusion Matrix: Hyperparameter Tuning')
plt.show()

# define entries of confusion matrix
tn_rf_tuned = cm_rf_tuned[0,0]
fn_rf_tuned = cm_rf_tuned[1,0]
tp_rf_tuned = cm_rf_tuned[1,1]
fp_rf_tuned = cm_rf_tuned[0,1]

# calculate the FPR and FNR
fpr_rf_tuned = fp_rf_tuned/(fp_rf_tuned+tn_rf_tuned)
fnr_rf_tuned = fn_rf_tuned/(fn_rf_tuned+tp_rf_tuned)
print(f"The FPR is {fpr_rf_tuned*100:.1f}% and the FNR is {fnr_rf_tuned*100:.1f}%.")
```

The confusion matrix indicates that our random forest model has a FPR of 75.6% and a FNR of 7.9%. This reflects the common tradeoff between recall and precision with imbalanced datasets. Since I optimized the model for recall of the minority class, the model now predicts more positive class labels in order to improve its recall. For this reason, only 7.9% of those who default are missed (false negatives) but this comes at the cost of incorrectly labeling 75.6% of those who don't default (false positives).

#### Step 6.3.2. Classification Report

The classification report is another common diagnostic tool used to evaluate the performance of a classification model, which summarizes four key metrics: precision, recall, f1-score, and support.

Precision quantifies how well you can trust a positive prediction, measured as the proportion of positive predictions that were correct:

$$
\begin{aligned}
\text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}}.
\end{aligned}
$$

The recall is complementary to precision, which quantifies the coverage of true positives measured as the proportion of true positives that were identified:

$$
\begin{aligned}
\text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}}.
\end{aligned}
$$

The F1-score is the harmonic mean of precision and recall, which offers a more balanced metric and can be particularly useful when there is a class imbalance:

$$
\begin{aligned}
\text{F1-score} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}.
\end{aligned}
$$

Lastly, support is the total number of occurrences in each class label and can help to provide context for the above key metrics, particularly in the case of a class imbalance.

Since the traditional classification report only includes F1-score (which weighs precision and recall equally), I will include an additional metric for F2-score which weighs recall more than precision. This is particularly useful for imbalanced datasets where recall is more important than precision.

```python
# compute classification report
classification_report_rf_tuned = classification_report(y_test,y_predict_rf_tuned)
f2_score_tuned = fbeta_score(y_test, y_predict_rf_tuned, beta=2)

print(classification_report_rf_tuned)
print(f"F2-score: {f2_score_tuned}")
```

The classification report reveals that the recall for the minority positive class is 92%, although its precision and F1-score are only 19% and 31%, respectively. I therefore successfully trained a random forest model to capture most of the rare individuals who default on their loans, but again this comes at the sacrifice of precision. This explains the relatively low F2 score: while the recall is quite high (92%), the precision is quite low (19%). The low precision and F2 score of the minority class means that there are a large number of individuals incorrectly flagged as being high risk for defaulting on their loans (i.e., false positives), but this is the expected tradeoff when dealing with imbalanced datasets: I chose to prioritize recall over precision.

#### Step 6.3.3. Precision-Recall AUC Score
The precision-recall (PR) curve is a diagnostic tool similar to the ROC curve that is particularly well-suited for imbalanced classification tasks. It visualizes the trade-off between precision (the proportion of correct positive predictions) and recall (the proportion of correctly-identified actual positives) across different classification thresholds. Unlike the ROC curve, which incorporates the FPR, the PR curve focuses solely on the performance of the the positive class. This makes it especially informative when the minority class is of primary interest, such as the case here, as it can provide a more realistic picture of model performance.

The Average Precision (AP) score, also referred to as the area under the precision-recall curve (PR-AUC), summarizes the trade-off between precision and recall into a single value. A high PR-AUC indicates that the model is able to maintain high precision without sacrificing recall across thresholds. In contrast, a low PR-AUC suggests that the model either misses many true positives (low recall) or makes many false positive predictions (low precision), or both.

```python
# compute precision-recall curve
precision_rf_tuned,recall_rf_tuned,thresholds_rf_tuned = precision_recall_curve(y_test,y_proba_rf_tuned)

# compute model's values
precision_final_rf_tuned = precision_score(y_test,y_predict_rf_tuned)
recall_final_rf_tuned  = recall_score(y_test,y_predict_rf_tuned)

# compute average precision
avg_precision_rf_tuned = average_precision_score(y_test,y_proba_rf_tuned)

# plot precision-recall curve
plt.figure(figsize=(8,6))
plt.plot(recall_rf_tuned,precision_rf_tuned,"-o",markersize=5,color="tab:blue",label=f'AP = {avg_precision_rf_tuned:.2f}')
plt.plot(recall_final_rf_tuned,precision_final_rf_tuned,'ro',markersize=5,label='Optimized threshold')
plt.axhline(0.16,linestyle="dashed",color="tab:orange",label=f"Baseline Precision = {y_test.mean():.2f}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve: Random Forest with Hyperparameter Tuning")
plt.xlim(-0.05,1.05)
plt.ylim(-0.05,1.05)
plt.legend()
plt.grid(True)
plt.show()

print(f"Optimized precision: {precision_final_rf_tuned:.2f}")
print(f"Optimized recall: {recall_final_rf_tuned:.2f}")
```

The precision-recall (PR) curve shows that the precision rapidly drops as with increasing recall, which is a common occurrence with imbalanced datasets. The threshold found to optimize the F2 score yields a precision of 19% for the minority class, which is only slightly higher than the baseline precision of 16%. The baseline precision is the percentage of individuals who defaulted on their loan and is equal to the probability of predicting a positive class with random guessing. While this precision is only slightly higher than the baseline precision, the recall is 92%, which is significantly higher than expected for random guessing. So while there are many false positives, the model successfully captures nearly all true positives. 

## Step 7. Random Forest with Hyperparameter Tuning and SMOTE Resampling
I will now incorporate SMOTE resampling to see if it improves recall. 

#### Step 7.1. Hyperparameter Tuning with SMOTE Resampling
To avoid data leakage, I will resample the training data using SMOTE within each $k$ fold. For this I will use `imblearn.pipeline`'s `Pipeline` function to ensure that SMOTE is only used on the training data within each fold. Since SMOTE uses distance metrics to interpolate between points, I will scale the data first using `StandardScaler()`. I will once again optimize on AP and use the same parameter grid for a fair comparison between models. Since I am already resampling with SMOTE, I won't force `class_weight='balanced'`. 

```python
# define pipeline: standard scaler, SMOTE resampling, then random forest model
pipeline = Pipeline(steps=[
    ('scaler',StandardScaler()),
    ('smote', SMOTE(random_state=42)),
    ('model', RandomForestClassifier(random_state=42))
])

# define parameter grid (same as previous step)
param_grid = {
    'model__n_estimators': [50, 100, 300, 500],
    'model__max_depth': [10, 20, 50, None],
    'model__min_samples_split': [2, 5, 10, 20],
    'model__min_samples_leaf': [10, 20, 50, 100, 200],
    'model__class_weight': [None,'balanced']
}

# random search optimizing recall
search_rf_smote = RandomizedSearchCV(estimator=pipeline,param_distributions=param_grid,n_iter=100,cv=cv,scoring='average_precision',random_state=42,n_jobs=-1,verbose=1)

# fit the model
search_rf_smote.fit(X_train,y_train)

# extract the model with the best recall
rf_tuned_smote = search_rf_smote.best_estimator_

# make predictions on the test set
y_predict_rf_tuned_smote = rf_tuned_smote.predict(X_test)

# print best-fit model parameters
print(f"best-fit model parameters (AP={search_rf_smote.best_score_:.2f}): {search_rf_smote.best_params_}")
```

Hyperparameter tuning yields a random forest model with a marginally higher optimized AP of 26% (compared to 27% previously). I will now turn to threshold optimization to improve recall.

### Step 7.2. Threshold Optimization
I will now identify the threshold that maximizes recall using the F2 score just like before. 

```python
# re-fit the model with the best-params model using the new training subset
rf_best_smote = search_rf_smote.best_estimator_
rf_best_smote.fit(X_train_tr,y_train_tr)

# get predicted probabilities
y_proba_rf = rf_best_smote.predict_proba(X_test_tr)[:, 1]

# iterate through thresholds
thresholds = np.linspace(0,1,100)

# collect F2 scores
F2_scores_smote = []
for threshold in thresholds:
    y_pred = (y_proba_rf >= threshold).astype(int)
    F2_score_smote = fbeta_score(y_test_tr, y_pred, beta=2)
    F2_scores_smote.append(F2_score_smote)

# find threshold that maximizes F2 score
idx_smote = np.argmax(F2_scores_smote)
best_threshold_rf_tuned_smote = thresholds[idx_smote]
best_F2_score_smote = F2_scores[idx_smote]

print(f"Best threshold: {best_threshold_rf_tuned_smote:.4f}, Optimal F2 score: {best_F2_score_smote:.3f}")
```

This yields an optimal threshold of 0.3636, which has a corresponding F2 score of 0.523. Interestingly, these are identical values as those obtained from the previous random forest model without SMOTE resampling. Next, I will train our final model on the entire training set and then tune the threshold to this optimal value. 

```python
# fit the model on the entire training set
rf_tuned_smote = search_rf_smote.best_estimator_
rf_tuned_smote.fit(X_train,y_train)

# make predictions on entire testing set
y_proba_rf_tuned_smote = rf_tuned_smote.predict_proba(X_test)[:,1]

# make final predictions using optimized threshold
y_predict_rf_tuned_smote = (y_proba_rf_tuned_smote>best_threshold_rf_tuned_smote).astype(int)
```

Let's now compare the class distribution for the previous model (without SMOTE) and the current model (with SMOTE). In particular, let's compare the number of correct predicted class labels to see whether SMOTE resampling led to more or less correct predictions for each class.

## Step 9. Summary and Next Steps
This demonstrates the process of building a random forest model using `scikit-learn`'s `RandomForestClassifier()` to predict whether an individual is likely to default on their loan based on financial information. An EDA revealed a significant class imbalance in the target variable (`not.fully.paid`), with those defaulting on their loans (positive class label) representing only 16% of the total data. Therefore, I carefully considered which performance metrics should be used to train the random forest model. 

Hyperparameter tuning was first used to optimize on average precision (AP), which is a common proxy for the area under the precision-recall curve (PR-AUC) and quantifies how well the model can discriminate between classes across all thresholds. This was used to identify the general class of models that best differentiate between those who are likely to default on their loan and those who are not. During hyperparameter tuning, I used `StratifiedKFold()` for cross-validation to ensure that all folds had a representative proportion of class labels and `RandomizedSearchCV()` to efficiently search through the large hyperparameter space. To penalize misclassifying defaulters more heavily, I set `class_weight="balanced"` to give greater weights to the minority class during the training phase.

Next, I optimized the classification threshold to prioritize the model for recall of the minority positive class (i.e., those who defaulted). Rather than optimizing simply on recall I used the F2-score, which is a variant of the F1-score that more heavily weights recall while still considering the precision. This yielded an optimal threshold of 0.36, as opposed to the default value of 0.5. 

The optimized random forest model achieved a recall of 92%, precision of 19%, F1-score of 31%, and F2-score of 52%. I therefore successfully trained a random forest model to capture most (92%) of the rare individuals who default on their loans but at the expense of precision, which is a common tradeoff when prioritizing recall with imbalanced datasets. The low precision, F1-score, and F2-score of the minority class means that there are a large number of individuals incorrectly flagged as being high risk for defaulting on their loans (i.e., false positives), but this is the expected tradeoff when choosing to prioritize recall over precision.

A secondary random forest model incorporating SMOTE resampling was used to handle the class imbalance. Since SMOTE uses a distance metric to interpolate between points, the data was first scaled with `StandardScaler()`. Using `imblearn.pipeline.Pipeline`, a pipeline was used to ensure that the data was scaled first and that SMOTE was only applied to the training data within each fold. However, this model had significantly worse recall (85% from 92%), only a marginally improved precision (20% from 19%) and F1-score (32% from 31%), and an unchanged F2-score. This means that SMOTE resampling marginally improved the model's ability to differentiate between those who are likely to default and those who aren't, but also led to quite a significant loss in its ability to identify those actually defaulted. As our priority was on recall, SMOTE resampling did not improve model performance in this case. The table below summarizes a comparison of these model's performance metrics.

Future work could involve other resampling techniques or alternative models that are better suited for imbalanced datasets, such as a balanced random forest model, in order to improve precision while maintaining recall.
